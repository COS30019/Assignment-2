```python
import pandas as pd
import folium

# Load new dataset
df = pd.read_csv('/Users/mouyly/Downloads/jupiter_notebook/latlong_map.csv')  # updated path to new file
print("Total rows:", len(df))
print("Unique Site IDs:", df['Site ID'].nunique())

# Check how many unique lat/lon pairs
print("Unique lat/lon positions:", df[['Lat', 'Long']].drop_duplicates().shape[0])

# Prepare unique site data
site_data = df[['Site ID', 'Lat', 'Long']].drop_duplicates()

# Define edge list (connections between Site IDs)
edges = [
    (2820, 3662),
    (3662, 3002),
    (3002, 3001),
    (4821, 3001),
    (4262, 3001),
    (4263, 3002),
    (3662, 4335),
    (3662, 4324),
    (4335, 4321),
    (4321, 4030),
    (4030, 4051),
    (4032, 4034),
    (4034, 4035),
    (4035, 3120),
    (3120, 3122),
    (3120, 4040),
    (4040, 3804),
    (4040, 3812),
    (4040, 4043),
    (4043, 4273),
    (970, 3685),
    (3685, 2000),
    (2000, 3682),
    (3682, 3126),
    (3126, 3127),
    (4273, 4043),
    (4030, 4032),
    (4030, 2825),
    (2825, 2827),
    (4051, 2827),
    (3180, 4051),
    (3180, 4057),
    (4057, 4063),
    (4063, 3127),
    (3122, 3127),
    (3127, 3126),
    (3804, 3812),
    (3804, 3682),
    (4040, 4272),
    (4270, 4272),
    (4262, 4263),
    (4264, 4266),
    (3001, 3662),
    (4363, 4264),
    (4264, 4270),
    (4264, 4324)
]

# Center the map around Boroondara (adjust if needed)
m = folium.Map(
    location=[-37.82, 145.06],
    zoom_start=13,
    tiles='cartodbpositron'
)

# Collect only nodes that are in edge list
connected_nodes = set()
for a, b in edges:
    connected_nodes.add(a)
    connected_nodes.add(b)

# Plot site markers in blue
for _, row in site_data[site_data['Site ID'].isin(connected_nodes)].iterrows():
    folium.CircleMarker(
        location=[row['Lat'], row['Long']],
        radius=5,
        color='blue',
        fill=True,
        fill_opacity=0.9,
        popup=f"Site ID: {row['Site ID']}"
    ).add_to(m)

# Draw green edges between connected site pairs
for a, b in edges:
    try:
        point_a = site_data[site_data['Site ID'] == a].iloc[0]
        point_b = site_data[site_data['Site ID'] == b].iloc[0]

        folium.PolyLine(
            locations=[
                [point_a['Lat'], point_a['Long']],
                [point_b['Lat'], point_b['Long']]
            ],
            color='green',
            weight=3,
            opacity=0.8
        ).add_to(m)
    except IndexError:
        print(f"⚠️ Skipping edge ({a}, {b}) — Site ID not found in latlong_map.csv")

# Save to HTML
m.save("manual_graph_map.html")
print("✅ Map saved as manual_graph_map.html — open it in your browser!")
```
