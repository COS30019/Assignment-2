```python
import pandas as pd
from sklearn.neighbors import NearestNeighbors
import networkx as nx
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv('Final_SCATS_with_LatLong.csv')
```

```python
print("Total rows:", len(df))
print("Unique Site IDs:", df['Site ID'].nunique())

# Check how many unique lat/lon pairs
print("Unique lat/lon positions:", df[['Lat', 'Long']].drop_duplicates().shape[0])
```

```python
import folium

#Prepare unique site data
site_data = df[['Site ID', 'Lat', 'Long']].drop_duplicates()

#define edge
edges = [
    (2000, 3682),
    (2000, 4043),
    (4043, 3120),
]

# center map Boroondara
m = folium.Map(
    location=[-37.82, 145.06],     # Melbourne area
    zoom_start=13,
    tiles='cartodbpositron'        #fir map style
)

#plot only node that appear in the connected edge
connected_nodes = set()
for a, b in edges:
    connected_nodes.add(a)
    connected_nodes.add(b)

#Add site markers blue
for _, row in site_data[site_data['Site ID'].isin(connected_nodes)].iterrows():
    folium.CircleMarker(
        location=[row['Lat'], row['Long']],
        radius=5,
        color='blue',
        fill=True,
        fill_opacity=0.9,
        popup=f"Site ID: {row['Site ID']}"
    ).add_to(m)

#Add green edges line
for a, b in edges:
    point_a = site_data[site_data['Site ID'] == a].iloc[0]
    point_b = site_data[site_data['Site ID'] == b].iloc[0]
    
    folium.PolyLine(
        locations=[
            [point_a['Lat'], point_a['Long']],
            [point_b['Lat'], point_b['Long']]
        ],
        color='green',
        weight=3,
        opacity=0.8
    ).add_to(m)

#Save map to HTML
m.save("map.html")
print("✅ Map saved as manual_graph_map.html — open it in your browser!")
```
