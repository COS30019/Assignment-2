```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## read data

```python
site_df = pd.read_csv('/Users/mouyly/Downloads/jupiter_notebook/SCATS_cleaned.csv' , skiprows=9)
site_df.head()
```

```python
# Drop columns that are completely empty (all NaN)
site_df = site_df.dropna(axis=1, how='all')

# Preview cleaned dataframe
site_df.head()
```

```python
print(site_df.columns)
```

## clean the data

### check for missing value

```python
site_df.isnull().sum()
```

```python
print(site_df.columns.tolist())
```

### check for duplicated 

```python
site_df.duplicated(subset=['Site Number']).sum()
```

```python
site_df[site_df['Site Number'].isnull()]
```

```python
# Remove duplicate site numbers, keeping the first occurrence
site_df = site_df.drop_duplicates(subset='Site Number', keep='first')

# Confirm it's done
print(f"Remaining rows after removing duplicates: {len(site_df)}")
```

## merging data

### with scats oct 2006

```python
#import scats 
long_df = pd.read_csv('Cleaned_SCATS_Oct2006.csv')
#rename site number
site_df.rename(columns={'Site Number': 'Site ID'}, inplace=True)
#merge
merged_df = pd.merge(long_df, site_df, on='Site ID', how='left')
#result
print("Shape after merge:", merged_df.shape)
print("Missing values:\n", merged_df.isnull().sum())
                    
```

```python
#save
merged_df.to_csv("Merged_SCATS_with_SiteListing.csv", index=False)
print(" File saved as 'Merged_SCATS_with_SiteListing.csv'")
```

```python
#load coordi
geo_df = pd.read_csv("Cleaned_SCATS_Oct2006_coord.csv")

# Merge location
final_df = pd.merge(merged_df, geo_df, on='Site ID', how='left')

# check for any unmatched coordinates
print("Missing after merge:\n", final_df[['Lat', 'Long']].isnull().sum())
```
